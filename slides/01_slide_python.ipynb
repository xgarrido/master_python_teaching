## **Structures conditionnelles**

Un bloc d'instructions lié à une condition se déclare avec les opérateurs `if/elif/else`

````python
is_python = True
is_cpp = False

if is_python:
    print("Cours de python")
elif is_cpp:
    print("Cours de C++")
else:
    print("Pas de cours d'informatique")
````

- la fin de chaque condition est matérialisée par **le caractère `:`**,
- **l'indentation (2 espaces ou une tabulation) délimite le bloc de condition**

Python offre la possibilité de former des *expressions* dont l'évaluation est soumise à une condition

````python
x = 2.0

y = x if x < 0 else x**2
y
````

````python
print("y est positif" if y > 0 else "y est négatif")
````

## **Répétition conditionnelle**

Pour répéter un bloc d'instructions tant qu'une condition est réalisée, Python propose la clause **`while`**

*Exemple:* Suite de Syracuse

````python
n = 27
while n != 1:
    if n % 2:
        n = 3 * n + 1
    else:
        n //= 2
    print(n, end=" ")
````

## **Boucles `for`**

Pour répéter un certain nombre de fois un bloc d'instructions, on utilisera la construction suivante

```python
for variable in objet:
    bloc d'instructions
```

*Exemple:*

````python
for c in "abcdef":
    print(c)
````

*Autres exemples:*

````python
for i in range(4):
    print(i)
````

````python
for i in range(-4, 4, 2):
    print(i)
````

## **Instructions `break` & `continue`**

Pour quitter une boucle `for` (ou `while`) en cours d'exécution, on utilisera l'instruction `break`

````python
for i in range(-4, 4, 2):
    if i == 0:
        break
    print(i)
````

S'il s'agit de passer outre le bloc d'instructions, on utilisera l'instruction `continue`

````python
for i in range(-4, 4, 2):
    if i == 0:
        continue
    print(i)
````

> L'instruction `continue` est particulièrement utile afin d'éviter une trop grande imbrication d'instructions `if` successives.
