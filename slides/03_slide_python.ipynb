# **Scripts/modules, programmation orientée object, librairie standard**

**Un script** est un fichier contenant un ensemble d'instructions Python. L'extension du fichier-script sera **`.py`** de telle sorte à ce que n'importe quel éditeur de texte ou environnement de développement intégré puisse mettre en oeuvre l'indentation, la coloration syntaxique ou d'autres fonctionnalités (complétion) en relation avec le langage.

*Exemple:* `citations.py`

```python
citation = "Une noisette, j'la casse entre mes fesses tu vois... JCVD"
for word in citation.split():
    print(word)
```

On pourra exécuter le script ci-dessus à l'aide la commande `python citations.py`.
