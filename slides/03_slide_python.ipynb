## **Scripts `python`**

**Un script** est un fichier contenant un ensemble d'instructions Python. L'extension du fichier-script sera **`.py`** de telle sorte à ce que n'importe quel éditeur de texte ou environnement de développement intégré puisse mettre en oeuvre l'indentation, la coloration syntaxique ou d'autres fonctionnalités (complétion) en relation avec le langage.

*Exemple:* `citations.py`

```python

for word in citation.split():
    print(word)
```

On pourra exécuter le script ci-dessus à l'aide la commande `python citations.py` ou dans une cellule `jupyter` (ainsi que dans l'interpréteur `ipython`) `%run citations.py`. 

## **Modules `python`**

À la différence d'un script, **un module `python`** est un fichier contenant un ensemble de fonctions/classes (voir programmation orientée objet) pouvant être utilisées par différents scripts

*Exemple:* `jcvd_collection.py`

```python
"""                                                                        
   A file with a lot of JCVD inside                                        
                                                                           
   This module holds several quotes from Jean-Claude Van Damme             
"""                                                                        
                                                                           
def quote0():                                                              
    print("Une noisette, j'la casse entre mes fesses tu vois...")          
                                                                           
def quote1():                                                              
    print("Quand tu prends confiance en la confiance tu deviens confiant.")
                                                                           
def quote2():                                                              
    print("Ce n'est pas moi qui parle...c'est nous qui parlons.")          
```

Pour pouvoir utiliser le module et ses fonctions, **il est nécessaire de l'importer** soit dans un script ou soit dans l'interpréteur

- Importation de base

````python
import jcvd_collection

jcvd_collection.quote1()
````

- Importation à l'aide d'un nom raccourci

````python
import jcvd_collection as jcvd

jcvd.quote1()
````

- Importation spécifique d'une fonction

````python
from jcvd_collection import quote1

quote1()
````

Par défaut, la localisation des modules se fait dans différents répertoires
  1) dans le répertoire local
  2) dans les répertoires définis au sein de la variable d'environnement `PYTHONPATH`
  3) dans l'ensemble des répertoires référencés par `sys.path`

````python
import sys

sys.path
````


## **Programmation orientée objet**

La programmation orientée objet (POO), ou programmation par objet, est *un paradigme de programmation informatique* élaboré par les Norvégiens Ole-Johan Dahl et Kristen Nygaard au début des années 1960 et poursuivi par les travaux d'Alan Kay dans les années 1970.

Il consiste en la définition et l'interaction de briques logicielles appelées objets; *un objet*
*représente un concept, une idée ou toute entité du monde physique, comme une voiture, une personne
ou encore une page d'un livre*.

Il possède *une structure interne et un comportement*, et il sait interagir avec ses pairs. Il s'agit donc de représenter ces objets et leurs relations; *l'interaction entre les objets via leurs
relations* permet de concevoir et réaliser les fonctionnalités attendues, de mieux résoudre le ou les problèmes.

Dès lors, l'étape de modélisation revêt une importance majeure et nécessaire pour la POO. C'est elle qui permet de transcrire les éléments du réel sous forme virtuelle.

*source [Wikipedia](https://fr.wikipedia.org/wiki/Programmation_orient%C3%A9e_objet)*

**Un objet** est une structure hébergeant des **données membres** (ou attributs) et des **fonctions membres** également appelées **méthodes**. La représentation sous forme d'objet est parfaitement adaptée à la programmation graphique (*GUI*) et à la description des détecteurs en physique (des particules/nucléaire).

Pour rappel, en Python tout est objet (variables, fonctions, classes).

*Exemple:* Déclaration d'un objet/classe `Student`

````python
class Student:
    def __init__(self, name):
        self.name = name

    def set_age(self, age):
        self.age = age

    def set_mark(self, mark):
        self.mark = mark
````

Dans cet exemple, les **données membres** sont `name`, `age` et `mark` tandis que les **méthodes** sont `__init__` (*i.e.* le constructeur de la classe), `set_age` et `set_mark`. 

*Exemple:* Création d'un objet de type `Student`

````python
student = Student("Patrick Puzo")
student.set_age(50)
student.set_mark(0.0)

print(f"Résultat de {student.name} : {student.mark}/20")
````

## **Gestions des exceptions**

Les exceptions sont la conséquence d'erreurs fonctionnelles

*Exemple:* erreur lors d'un résultat indéfini

````python
0/0
````

*Exemple:* erreur typographique dans le nom d'une fonction, d'un attribut

````python
import sys

sys.pth
````

Pour "attraper" les exceptions avant qu'elles ne causent l'arrêt du programme, on utilise les instructions `try/except`

````python
while True:
    try:
        x = int(input("Veuillez saisir un nombre: "))
        break
    except ValueError:
        print("Je crois avoir demandé un nombre !")
````

Pour "lever" une exception, on utilise l'instruction **`raise`**...

````python
def achilles_arrow(x):
    if abs(x - 1) < 1e-3:
        raise StopIteration
    x = 1 - (1 - x) / 2.0
    return x
````

... pour mieux pouvoir la récupérer dans un second bloc

````python
x = 0
while True:
    try:
        x = achilles_arrow(x)
    except StopIteration:
        break
````

L'écriture dans un fichier se fait nécessairement par le biais de chaîne de caractères

````python
f = open("/tmp/QA.txt", "w")
f.write("Qu'est qu'un chat qui voit dans le futur ?")
f.close()
````

La lecture dans un fichier peut se faire de la façon suivante...

````python
f = open("/tmp/QA.txt", "r")
s = f.read()
print(s)
f.close()
````

...ou en lisant le fichier ligne par ligne

````python
with open("/tmp/QA.txt", "r") as f:
    for line in f:
        print(line)
````

L'instruction `with` assure que le fichier sera fermé quoiqu'il advienne notamment si une exception est levée. On priviligiera donc cette utilisation y compris et surtout pour l'écriture d'un fichier

````python
with open("/tmp/QA.txt", "w") as f:
    f.write("Qu'est qu'un chat qui voit dans le futur ?")
````
