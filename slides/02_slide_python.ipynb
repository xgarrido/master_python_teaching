# **&nbsp;*Containers* & autres fonctions**

## **Les séquences ou *containers*&nbsp;**

En plus des types fondamentaux, Python propose nativement un ensemble d'objets à accès séquentiel dont :
- les chaînes de caractères,
- les listes et *tuples*,
- les dictionnaires.

### **Format et *Formatage*&nbsp;**

- Utilisation de la méthode `format` (*cf.* [Python 3 string format](https://docs.python.org/3/library/string.html#new-string-formatting))

```python
canevas = "Nom: {}, prénom: {}, date de naissance: {}"

canevas.format("Van Rossum", "Guido", "31/01/1956")
```

```python
canevas = "Nom: {nom}, prénom: {prenom}, date de naissance: {date}"

canevas.format(date="31/01/1956", nom="Van Rossum", prenom="Guido")
```

```python
canevas = "L'année {0:d} s'écrit {0:b} en binaire et {0:x} en héxadécimal"

canevas.format(2026)
```

- Utilisation de ***f-string***

```python
nom, prenom, age = "Van Rossum", "Guido", 64
print(f"Nom: {nom}, prénom: {prenom}, age: {age}")
```

## **Les listes**

Une liste est un objet qui permet de stocker **une collection d'objets de tous types**

- Initialisation d'une liste

```python
bazar = []

bazar = ["rouge", "vert", "bleu", "noir"]

bazar = [1, 2, 3, 4]

bazar = [1, 2, "bleu", 3, 4]

type(bazar)
```

- Conversion en liste

```python
bazar = list(range(4))
bazar
```

```python
bazar = list("abcdef")
bazar
```

- Initialisation d'une liste « en compréhension »

```python
bazar = [x**2 for x in range(4)]
bazar
```

```python
[x**2 for x in range(1, 100) if x % 10 == 3]
```

### **Manipulation d'une liste**

Comme pour les chaînes de caractères, l'accès aux éléments d'une liste se fait par indice

```python
kebab = ["salade", "tomates", "oignons", "sauce blanche"]

kebab[2]
```

```python
kebab[-1]
```

```python
kebab[1:3]
```

```python
kebab[0] = "sans salade"
kebab
```

> À la différence des chaînes de caractères, les listes sont des objets **mutables** *i.e.* modifiables !

On peut également ajouter ou supprimer des éléments *via* des méthodes associées

```python
kebab = ["salade", "tomates", "oignons", "sauce blanche"]

kebab.append("frites")  # équivalent à kebab += ["frites"]
kebab
```

```python
kebab.pop()
```

```python
kebab
```

```python
kebab.extend(["frites", "coca"])
kebab
```

```python
kebab.insert(3, "harissa")
kebab
```

Pour parcourir les éléments d'une liste par le biais d'une boucle `for`, on écrit

```python
for item in kebab:
    print(item)
```

Si on souhaite, en plus de l'élément, accéder à l'indice au sein de la liste, on procède soit à l'aide de la fonction `range`

```python
for idx in range(len(kebab)):
    print(idx, kebab[idx])
```

soit, plus élégamment, à l'aide de la fonction `enumerate`

```python
for idx, item in enumerate(kebab):
    print(idx, item)
```

Il existe d'autres méthodes/fonction associées à la manipulation des listes. Comme pour les chaînes de caractères, l'ensemble des méthodes associées aux objets de type `list` sont accessibles *via* l'aide en ligne *i.e.* `help(list)` ou en utilisant la complétion `kebab.<TAB>`.

## **Les *tuples*&nbsp;**

Un *tuple* correspond à **une liste *immutable*** *i.e.* non modifiable

```python
kebab = ("salade", "tomates", "oignons", "sauce blanche")

kebab
```

```python
type(kebab)
```

```python
kebab[0] = "saucisson"
```

## **Les dictionnaires**

Les dictionnaires sont des structures **mutables**, ordonnées et formées d'enregistrements de type **`clé:valeur`**. Le seul moyen d'accéder à une valeur particulière est par l'intermédiaire de sa clé et non plus par un indice entier.

```python
tel = {"jacquie": 5752, "michel": 5578}
tel
```

```python
tel["adrien"] = 5915
tel
```

```python
tel["michel"]
```

```python
tel.keys()
```

```python
tel.values()
```

```python
"adrien" in tel
```

```python
for key, value in tel.items():
    print(f"Clé/Valeur : {key.capitalize()}/{value}")
```

## **Les fonctions**

- Une fonction est **un bloc d'instructions** qui a reçu **un nom**
- Une fonction peut :
    1) dépendre d'un certain nombre de paramètres $\to$ **les arguments**
    2) renvoyer un résultat au moyen de l'instruction **`return`**

- Quelques fonctions intégrées au langage Python
    - `help` : aide interactive sur un nom $\to$ `help(float)`
    - `input` : entrée au clavier $\to$ `n = int(input("N ?"))`
    - `print` : affiche à l'écran $\to$ `print(n)`
    - `type`, `sum`, `range`, `min/max`, ...

### **Déclaration de fonction**

- Fonction sans argument et sans valeur de retour

```python
def dummy():
    print("Fonction `dummy`")


dummy()
```

> Par défaut, la valeur de retour d'une fonction est `None`

- Fonction **avec argument et valeur de retour**

```python
def aire_disque(rayon):
    return 3.14 * rayon**2


aire_disque(1.5)
```

```python
aire_disque()
```

- Fonction **avec argument par défaut et valeur de retour**

```python
def aire_disque(rayon=10.0):
    return 3.14 * rayon**2


aire_disque(1.5)
```

```python
aire_disque()
```

```python
aire_disque(rayon=20)
```

```python
kwargs = {"rayon": 20}
aire_disque(**kwargs)
```

- Grâce à l'affectation en parallèle, une fonction peut **retourner plusieurs valeurs**

```python
def decomposer(entier, diviseur):
    return entier // diviseur, entier % diviseur


partie_entiere, reste = decomposer(20, 3)
partie_entiere, reste
```

### **Fonctions $\lambda$** 

Les fonctions $\lambda$ sont des fonctions dites **anonymes** *i.e.* sans nom pouvant être appliquée "à la volée" dans une expression

```python
f = lambda x: x**2
f(2)
```

```python
g = lambda x, y, z: 100 * x + 10 * y + z
g(1, 2, 3)
```

### **Documentation de fonctions**

La documentation des fonctions en Python est intrinsèquement liée au code et peut, sans difficulté, être implémentée conjointement au code informatique lui-même

```python
def dummy():
    """Cette fonction ne sert strictement à rien.

    En plus détaillé, cette fonction ne sert toujours
    à rien mais la description est plus longue.
    """


help(dummy)
```

> Pour plus de détails sur les us et coutumes en matière de documentation *cf.* [*Docstrings convention*](https://www.python.org/dev/peps/pep-0257)

### **Fonctions objets**

Les fonctions sont des objets ce qui implique qu'elles peuvent être :
1) affectées à une variable
2) un élément dans une séquence (liste, dictionnaires,...)
3) passées comme argument à une autre fonction

```python
ad = aire_disque

ad(2)
```

```python
table = {"Calcul de l'air d'un disque": ad}
table["Calcul de l'air d'un disque"]()
```

```python
def func2func(fcn):
    return fcn()


func2func(ad)
```
